## Завдання 1
Побудуйте дерево рішень і перевірте, наскільки добре воно передбачає ціни на житло.
- Завантажте датасет California Housing.
- Використовуйте ознаки MedInc, HouseAge, AveRooms, AveBedrms.
- Розділіть дані на train/test (80/20, random_state=42).
- Навчіть модель `DecisionTreeRegressor(max_depth=4, random_state=42)`.
- Обчисліть R² на тестовій вибірці.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score

# Завантаження даних
california = fetch_california_housing()
X = pd.DataFrame(california.data, columns=california.feature_names)
y = california.target

# Відбір ознак
features = ['MedInc', 'HouseAge', 'AveRooms', 'AveBedrms']
X_subset = X[features]

# Розділення на train/test
X_train, X_test, y_train, y_test = train_test_split(X_subset, y, test_size=0.2, random_state=42)

# Навчання моделі
dt_regressor = DecisionTreeRegressor(max_depth=4, random_state=42)
dt_regressor.fit(X_train, y_train)

# Прогноз та обчислення R²
y_pred_dt = dt_regressor.predict(X_test)
r2_dt = r2_score(y_test, y_pred_dt)
print(f'DecisionTreeRegressor R²: {r2_dt:.4f}')
```

Порівняйте якість передбачень одного дерева й ансамблю дерев.
Використовуйте ті самі дані й ознаки.
- Навчіть: `DecisionTreeRegressor(max_depth=4, random_state=42)` та `RandomForestRegressor(n_estimators=100, random_state=42)`.

```python
from sklearn.ensemble import RandomForestRegressor
from IPython.display import display

# Модель випадкового лісу
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=42)
rf_regressor.fit(X_train, y_train)

# Прогноз та R²
y_pred_rf = rf_regressor.predict(X_test)
r2_rf = r2_score(y_test, y_pred_rf)

# Оформлення результату у таблиці
results_df = pd.DataFrame({
    'Модель': ['DecisionTreeRegressor', 'RandomForestRegressor'],
    'R²': [r2_dt, r2_rf]
})
display(results_df)
```

## Завдання 4
Спробуйте розв'язати задачу класифікації на основі цін на житло.
- Використовуйте California Housing.
- Перетворіть цільову змінну на бінарну: 1, якщо ціна вища за медіану, 0 — якщо нижча або дорівнює.
- Розділіть дані на train/test (80/20, random_state=42).
- Навчіть модель `LogisticRegression(max_iter=1000)`.
- Побудуйте confusion matrix і виведіть Accuracy, Precision, Recall, F1.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
import matplotlib.pyplot as plt
import seaborn as sns

# Перетворення цільової змінної в бінарну
median_price = np.median(y)
y_class = (y > median_price).astype(int)

# Розділення даних на train/test (використовуємо всі ознаки)
# Розділяємо масив наново за умовами
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X, y_class, test_size=0.2, random_state=42)

# Навчання моделі логістичної регресії
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_c, y_train_c)

# Прогноз
y_pred_c = log_reg.predict(X_test_c)

# Обчислення метрик
conf_matrix = confusion_matrix(y_test_c, y_pred_c)
accuracy = accuracy_score(y_test_c, y_pred_c)
precision = precision_score(y_test_c, y_pred_c)
recall = recall_score(y_test_c, y_pred_c)
f1 = f1_score(y_test_c, y_pred_c)

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")

# Побудова Confusion Matrix
plt.figure(figsize=(6, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=['<= Median', '> Median'], yticklabels=['<= Median', '> Median'])
plt.title('Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```
