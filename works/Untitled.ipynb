```python
import os
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Path to dataset root (update this to your dataset location)
DATASET_DIR = "D:/chest_xray/"

# Image transformations: resize, convert to tensor, normalize
transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Resize to match model input size
    transforms.ToTensor(),          # Convert to PyTorch tensor
    transforms.Normalize(mean=[0.485, 0.456, 0.406],  # ImageNet mean
                         std=[0.229, 0.224, 0.225])   # ImageNet std
])

# Load datasets
train_dataset = datasets.ImageFolder(os.path.join(DATASET_DIR, "train"), transform=transform)
val_dataset   = datasets.ImageFolder(os.path.join(DATASET_DIR, "val"), transform=transform)
test_dataset  = datasets.ImageFolder(os.path.join(DATASET_DIR, "test"), transform=transform)

# Create DataLoaders
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=4)
val_loader   = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=4)
test_loader  = DataLoader(test_dataset, batch_size=32, shuffle=False, num_workers=4)

# Check class names
print("Classes:", train_dataset.classes)  # ['NORMAL', 'PNEUMONIA']

# Example: iterate over one batch
images, labels = next(iter(train_loader))
print(f"Batch image tensor shape: {images.shape}")
print(f"Batch labels: {labels}")
```
